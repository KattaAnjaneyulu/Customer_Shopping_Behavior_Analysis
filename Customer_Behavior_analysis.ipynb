```python
df= pd.read_csv('customer_shopping_behavior.csv',na_values=['NA','N/A'])  

# When reading the csv file ,treat NA,N/A as missing values.
# Without na_values 'NA','N/A' in data are treated as text.
```

```python
df.isnull().sum()
```

```python
df.isna().any()
```

```python
df= df.rename(
    columns={'Item Purchased':'Item_Purchased'})  #renaming column name 
```

```python
df[['Item_Purchased','Category']] # selectind and viewing two columns 
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()    # making the column names to lower case to make it to snake case 
df.columns = df.columns.str.replace(' ','_')  # replacing spaces with _ with the help of string fuction 
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
# Feature engineering 
# Create a Column age_group
labels = ['Young Adult','Adult','Middle_aged','Senior']
df['age_group'] = pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head(5)
```

```python
df.head(5)
```

```python
# Create column purchase_frequency_dates
frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Monthly' : 30,
    'Quarterly' : 90,
    'Bi-Weekly' : 14,
    'Annually' : 365,
    'Every 3 Months' : 90
 }

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['frequency_of_purchases','purchase_frequency_days']].head(6)
```
